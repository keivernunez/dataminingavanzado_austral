# **Árboles de Decisión**

Los árboles de decisión son modelos predictivos que se construyen mediante reglas binarias (utilizando SI/NO, <= o >, etc.) que dividen los casos de un conjunto de datos, usando los atributos de dichos casos como principales elementos para realizar estas divisiones.

En muchos casos, un único modelo global no es suficiente para identificar una solución aplicable a un espacio muestral, ya que las funciones que definen estos espacios muestrales pueden ser muy complejas. Por lo tanto, existen alternativas en el _machine learning_ que permiten descomponer estos problemas y generar predictores que interactúen de manera adecuada en función de la relación existente entre las variables.

Los árboles de decisión pueden ser utilizados para resolver problemas tanto de regresión (cuando la variable objetivo es continua) como de clasificación (cuando la variable objetivo es discreta, generalmente dicotómica).

**Ventajas de los árboles**

- Los árboles son fáciles de interpretar.
- Los modelos que se basan en un solo árbol son graficables, lo cual facilita la explicación de los mismos.
- Al no ser un modelo paramétrico (o estadístico), no requiere cumplir condiciones previas para ser implementado.
- No son muy influenciados por los _outliers_.
- Se utilizan mucho en la exploración de los datos, ya que, al ser fáciles de interpretar, se pueden obtener muchos _insights_ rápidamente.

**Desventajas de los árboles**

- Una de las principales desventajas es el bajo poder predictivo que tienen los modelos basados en un solo árbol. Esto se debe a su alta capacidad de hacer _overfitting_ (alto sesgo) y a la gran varianza que presentan.
- Son sensibles a conjuntos de datos desbalanceados (cuando un valor de la variable objetivo es mucho más frecuente que los otros).

### _**Overfitting y Underfitting, ¿qué significan estos términos?**_

Un mal resultado en un modelo de _machine learning_ puede deberse a dos conceptos importantes: _overfitting_ (sobreajuste) y _underfitting_ (subajuste). Estos son aspectos cruciales a considerar al construir un modelo predictivo. Al entrenar un modelo, utilizamos datos que se supone son una muestra representativa del universo total (tanto histórico como futuro). En base a esto, creamos una función que se adapte lo mejor posible a los datos históricos para que luego pueda predecir comportamientos futuros.

El _overfitting_ y el _underfitting_ hacen referencia a fallos del modelo al intentar "ajustar" los datos y generalizar el conocimiento que se espera que adquiera.

Los modelos de _machine learning_ deben ser capaces de generalizar los conceptos. Por ejemplo, si vemos un perro labrador por primera vez y alguien nos dice: "_esto es un perro_", y luego nos muestran otro perro (por ejemplo, un maltés) y nos preguntan: "_¿esto es un perro?_", podríamos decir "_No_". Pero si en vez de un solo perro hubiéramos visto fotos de más de 10 razas, estaríamos en mejores condiciones de identificar una nueva raza como un perro.

Entrenar modelos computacionales requiere un conjunto de datos de entrada que permita generalizar un concepto para que, al ser consultado, el modelo pueda dar una respuesta certera.

Resolver el problema del sesgo y la varianza es clave para abordar el _overfitting_ y el _underfitting_. El sesgo disminuye y la varianza aumenta en relación a la complejidad del modelo. A medida que se agregan más parámetros, la complejidad aumenta, y la varianza se convierte en la principal preocupación, mientras que el sesgo disminuye.

**¿Qué es el Sesgo?**

El sesgo mide qué tan bien el modelo se ajusta a los datos de entrenamiento. Un sesgo bajo significa que la función encontrada por el modelo se ajusta bien a los datos de entrenamiento, mientras que un sesgo alto indica lo contrario.

**¿Y la Varianza?**

La varianza mide la desviación de las predicciones. Indica cuánto cambiarán los resultados de las predicciones si los datos varían. En un modelo bien ajustado, cambiar el conjunto de datos no debería afectar significativamente las predicciones. Si el modelo tiene alta varianza, las predicciones se verán afectadas por los valores del conjunto de datos.

El _overfitting_ ocurre cuando un modelo es excesivamente complejo. Un modelo con _overfitting_ muestra un rendimiento excelente en el entrenamiento, pero al validar con nuevos datos, los resultados empiezan a fluctuar.   
El _underfitting_ ocurre cuando un algoritmo no logra capturar la tendencia subyacente de los datos. En este caso, el modelo no puede ajustarse correctamente a los datos de entrenamiento ni generalizar a nuevos datos.

### **Tipos de algoritmos de árboles**

### Árboles de Clasificación

Imaginemos una gran caja de frutas variadas, como manzanas, naranjas y plátanos. El objetivo es crear un sistema automatizado de clasificación. Un árbol de clasificación funciona como una herramienta visual que ayuda a identificar cada fruta haciendo preguntas sobre sus características, como el color, el tamaño y la forma. Basado en las respuestas, el árbol sigue diferentes caminos para determinar el tipo de fruta en cuestión. Por ejemplo: "Si es roja y redonda, es una manzana; si es alargada y anaranjada, es un plátano". En términos simples, se puede comparar con el juego "20 Preguntas" para descubrir qué tipo de fruta es.

### Árboles de Decisión

Los árboles de decisión actúan como herramientas para la toma de decisiones, similares a mapas que ayudan a navegar por las opciones. Por ejemplo, al planificar un fin de semana, una decisión crucial podría ser: "_¿Debería quedarme en casa o salir?_". Dependiendo de la respuesta, el árbol guía hacia decisiones más precisas, como elegir entre ver una película o leer un libro si se decide quedarse en casa. Este proceso continúa hasta llegar a un resultado al final de una rama, ofreciendo orientación clara sobre qué curso de acción seguir. Así, un árbol de decisión permite desglosar decisiones complejas en etapas más manejables, similar a seguir una ruta instruccional.

### Árboles de Regresión (CART)

Los árboles de regresión representan un enfoque sofisticado para predecir valores numéricos. Por ejemplo, al intentar estimar el costo de una casa usando factores como su tamaño y ubicación, los árboles de regresión analizan datos de múltiples casas y sus precios para identificar patrones. Al hacer preguntas específicas como "_¿La casa supera los 2000 metros cuadrados?_" o "_¿Está situada en una zona determinada?_", el árbol puede generar estimaciones precisas para los precios basándose en similitudes con otras propiedades. En esencia, este método implica obtener información a partir de instancias previas para hacer predicciones informadas sobre valores numéricos.

### El índice de GINI

El coeficiente de Gini es una medida que se utiliza para calcular la homogeneidad, indicando la desigualdad o impureza en un conjunto de datos. Se usa comúnmente en algoritmos de árboles de decisión, incluyendo árboles de regresión. En este contexto, el coeficiente Gini evalúa qué tan bien una partición separa los datos en diferentes clases o grupos, midiendo la impureza de una partición al calcular la probabilidad de que un punto de datos elegido al azar sea mal clasificado con base en la distribución por clase en cada subconjunto creado por la división. Un coeficiente Gini de 0 indica pureza perfecta, es decir, que todos los puntos de datos dentro del subconjunto pertenecen a la misma clase. Un coeficiente Gini de 1 indica máxima impureza, lo cual significa que los puntos de datos dentro del subconjunto están distribuidos uniformemente entre diferentes clases.

En los árboles de regresión, el coeficiente Gini se usa como criterio para evaluar las posibles divisiones en los datos. Al decidir qué característica dividir y dónde hacer la división, el algoritmo del árbol de regresión elige la división que resulte en el menor coeficiente Gini, indicando la mayor reducción en impureza o error de predicción.

$$GINI(t)= 1-{\sum_{j=1}^n P(j|t)^2}$$
Donde:
- $P(j|t)$ son las probabilidades de las clases.

Y el GINI de la división de un nodo es:

$$GINI_{SPLIT}= \sum_{i=1}^n \frac{N_i}{N} GINI(i)$$

Donde:
- $N_i$ son los casos que quedan en el nodo $i$.
- $N$ es el total de casos del nodo padre.
- $GINI(i)$ es el índice de GINI del nuevo nodo.

**_Entropía - Ganancia de Información_**

La entropía mide la incertidumbre de una fuente de información. Otra forma de verla es como la cantidad de información promedio que poseen los símbolos utilizados.

$$Entropía(t) = - \sum_{j=0}^n p(j|t) log_2  p(j|t)$$

Donde:
- $p(j|t)$ son las probabilidades de las clases.

Para calcular la ganancia de información debemos realizar lo siguiente:

$$GAIN_{SPLIT} = Entropía(p) - \left(\sum_{i=1}^k \frac{n_i}{n} Entropía(i)\right)$$

Donde:
- $k$ es la cantidad de nodos que se desprenden del nodo padre.
- $n_i$ es la cantidad de observaciones en cada uno de los nodos hijos.
- $n$ es la cantidad total de observaciones del nodo padre.
- $Entropía(i)$ es la entropía del nodo hijo $i$.

Los árboles eligen la mejor partición según maximice la ganancia de información. Para evitar pequeñas particiones se puede utilizar una penalización, conocida como _Gain Ratio_ o razón de ganancia, que se calcula de la siguiente forma:

$$Gain_{ratio} = \frac{GAIN_{split}}{Split_{info}}$$

Donde:

- $Split_{info} = -\sum_{i=1}^k \frac{n_i}{n}log\frac{n_i}{n}$
- $k$ es la cantidad de nodos que se desprenden del nodo padre.
- $n_i$ es la cantidad de observaciones en cada uno de los nodos hijos.
- $n$ es la cantidad total de observaciones del nodo padre.

### Árboles de Regresión

Este tipo de modelos se aplican cuando la variable a predecir (variable objetivo) es continua. El funcionamiento del árbol consiste en que las diferentes observaciones de un conjunto de datos (también conocidas como registros) se bifurcan en el árbol mediante los valores de sus atributos, llegando a un nodo terminal. La predicción del árbol será la media de los valores de las observaciones ubicadas en cada uno de los nodos. Por lo tanto, cuando un nuevo valor entra en el modelo, pasará por cada una de las ramas dependiendo de los valores de sus atributos y se alojará en un nodo terminal, tomando como predicción la media del nodo.

#### ¿Cómo se entrena un árbol?

De manera simple, los árboles (tanto de regresión como de clasificación) se entrenan en dos etapas:

1) Se realiza una división sucesiva del espacio de los predictores, generando regiones no solapantes (mutuamente excluyentes) denominadas nodos terminales. Esto significa que una observación no puede estar en dos nodos al mismo tiempo.

2) Se predice la variable respuesta en cada una de las regiones (nodos terminales).

Aunque estas dos etapas parecen simples, es necesario definir una metodología para crear estas regiones, es decir, determinar qué predictores son los mejores (atributos de cada observación) y cuál es el punto de corte óptimo (valor que mejora dicho corte). Aquí radica la diferencia entre los árboles de regresión y los de clasificación.  
En los árboles de regresión, el método más utilizado para identificar estas divisiones es el _Residual Sum of Squares_ (suma de residuos al cuadrado, donde el residuo es la diferencia entre el valor real y el predicho). El objetivo es encontrar las $J$ regiones que minimicen esta suma:

$$RSS = \sum_{i=1}^{J}\sum_{i∈R_j} (y_i - \hat{y_{R_i}})^2$$

Donde $y_{R_i}$ es la media de la región $R_j$.  
En otras palabras, se busca una distribución de regiones donde la suma de las desviaciones al cuadrado entre las observaciones y la media de la región a la que pertenecen sea lo menor posible. Dado que no se pueden considerar todas las posibles particiones, se utiliza un método denominado _Recursive Binary Splitting_ o división binaria recursiva.

#### Recursive Binary Splitting

El objetivo del método es encontrar en cada iteración el predictor $x_j$ y el punto de corte $s$ (o umbral) tal que, al distribuir las observaciones en las regiones ${X|Xj<s}$ y ${X|Xj≥s}$, se logre la mayor reducción posible en el RSS.

El algoritmo sigue los siguientes pasos:

1) El proceso comienza en la raíz del árbol, donde todas las observaciones pertenecen a la misma región.

2) Se identifican todos los posibles puntos de corte (umbrales) $s$ para cada uno de los predictores $(X_1, X_2,…, X_p)$. En el caso de predictores cualitativos, los posibles puntos de corte son cada uno de sus niveles. Para predictores continuos, se ordenan de menor a mayor sus valores, y el punto intermedio entre cada par de valores se utiliza como punto de corte.

3) Se calcula el RSS total que se consigue con cada posible división identificada en el paso 2.
$$RSS = RSS_{split 1}+RSS_{split 2}$$

4) Se selecciona el predictor $X_j$ y el punto de corte $s$ que resultan en el menor RSS total, es decir, que generan las divisiones más homogéneas posibles. Si existen dos o más divisiones que consiguen la misma mejora, la elección entre ellas es aleatoria.

5) Se repiten iterativamente los pasos 1 a 4 para cada una de las regiones creadas en la iteración anterior, hasta que se alcanza algún criterio de parada. Algunos de los criterios más comunes son: que ninguna región contenga menos de $n$ observaciones, que el árbol tenga un máximo de nodos terminales, o que la incorporación del nodo reduzca el error en al menos un % mínimo.

Esta metodología tiene dos implicancias:

- Cada división óptima se identifica en función del impacto que tiene en ese momento, sin considerar si es la mejor división para futuras divisiones del árbol.

- En cada división se evalúa un único predictor con preguntas binarias (sí/no), lo que genera dos nuevas ramas del árbol por cada división. Aunque es posible evaluar divisiones más complejas, hacer una pregunta sobre múltiples variables a la vez es equivalente a hacer múltiples preguntas sobre variables individuales.

Nota: Los algoritmos que implementan _Recursive Binary Splitting_ suelen incorporar estrategias para evitar evaluar todos los posibles puntos de corte. Por ejemplo, para predictores continuos, primero se crea un histograma que agrupa los valores y luego se evalúan los puntos de corte en cada región del histograma $(bin)$.

### Árboles de Clasificación

Para la construcción de este tipo de modelo se utiliza el mismo método de _Recursive Binary Splitting_ mencionado anteriormente. El único cambio es en la función que se optimiza. Existen varias funciones objetivo, cuyo objetivo es obtener nodos lo más puros/homogéneos posibles.  
Las funciones más utilizadas son:

**_Classification Error Rate_**: Es la proporción de observaciones que no pertenecen a la clase más común del nodo.

**Función de costo para la elección de la mejor partición**

Para determinar la mejor partición, se define una función de costo que asigna un puntaje al nodo padre, utilizando los promedios ponderados de los índices individuales de sus nodos hijos.

La función de costo se calcula multiplicando el porcentaje de casos que cayó en cada uno de los nodos por la métrica de impureza correspondiente. Luego se suman las métricas de impureza para obtener la función de costo total.

Independientemente de la medida que se utilice para buscar el punto óptimo, el algoritmo funciona de la siguiente forma:

1) Para cada posible división, se calcula el valor de la medida en los nodos resultantes.

2) Se suman los dos valores, ponderando cada nodo por la fracción de observaciones que posee.

3) La división con menor o mayor valor (dependiendo de la métrica empleada) se selecciona como punto de corte óptimo.

Las métricas más utilizadas son Gini y Entropía, ya que son las más sensibles a la homogeneidad de los nodos.

# **Ensembles**

Los modelos estadísticos y de _Machine Learning_ enfrentan un problema fundamental: el equilibrio entre el sesgo (_bias_) y la varianza (_variance_).  
El concepto de **sesgo** se refiere a cuánto, en promedio, se desvían las predicciones de los valores reales. En otras palabras, indica cuán precisa es la relación encontrada entre los predictores y la variable respuesta.  
Por otro lado, la **varianza** mide cuánto cambia el modelo dependiendo de los datos utilizados en el entrenamiento. Lo ideal es que un modelo pueda _**generalizar**_ su comportamiento, es decir, que sus resultados no varíen significativamente al ser evaluado con nuevas observaciones diferentes a las utilizadas en el entrenamiento.  
A medida que aumenta la complejidad del modelo, este logra adaptarse mejor a la realidad que se intenta modelar. Sin embargo, si esta complejidad no se controla, puede aparecer un fenómeno denominado _overfitting_, que ocurre cuando el modelo se ajusta tanto a los datos de entrenamiento que no es capaz de identificar correctamente nuevas observaciones. Por lo tanto, el mejor modelo será aquel que pueda equilibrar de la mejor manera el sesgo y la varianza.  
La gran pregunta es: ¿cómo controlar el sesgo y la varianza? Los árboles pequeños tienen poca varianza pero un alto sesgo, lo que significa que representan bien la relación entre las variables, pero pueden no captar toda la complejidad de los datos. Por el contrario, los árboles grandes se ajustan mucho a los datos de entrenamiento, resultando en bajo sesgo pero alta varianza. Un método para resolver esta problemática es trabajar con métodos de _ensembles_.

Los **ensembles** son métodos que combinan múltiples modelos en uno solo con el objetivo de lograr un equilibrio entre el sesgo y la varianza. Los dos _ensembles_ más utilizados son:

- **Bagging**: Se generan muchos modelos, cada uno con un subconjunto distinto de datos de entrenamiento. Todos los modelos contribuyen al veredicto final, ya sea promediando las predicciones (en variables continuas) o seleccionando la clase más frecuente (en variables categóricas). El modelo más conocido de esta técnica es el **Random Forest**.
- **Boosting**: Se ajustan secuencialmente modelos sencillos, denominados _weak learners_. El valor final es la media de las predicciones en variables continuas o la clase más votada en variables categóricas.

Ambos métodos buscan optimizar el equilibrio entre el sesgo y la varianza, pero lo hacen de manera diferente:
- En _bagging_, se utilizan modelos con poco sesgo pero alta varianza; al combinar estos modelos, se logra reducir la varianza sin afectar demasiado el sesgo.
- En _boosting_, se emplean modelos con baja varianza pero alto sesgo; al ajustarlos secuencialmente, se logra reducir el sesgo.

En _bagging_, las muestras aleatorias se toman utilizando la técnica de _bootstrap_. En _boosting_, los pesos de cada modelo se ajustan en cada iteración según los errores cometidos.

Estos tipos de modelos suelen ofrecer un rendimiento superior en comparación con modelos más clásicos (como las regresiones logísticas), aunque a costa de perder algo de interpretabilidad.

## **Bagging** (Bootstrap Aggregation)

Dadas $n$ muestras de observaciones independientes $Z_1, ..., Z_n$, cada una con una varianza $\epsilon^2$, la varianza de la media de las observaciones es $\epsilon^2/n$. En otras palabras, al promediar un conjunto de observaciones, se reduce la varianza.  
El _Random Forest_ es una implementación del método _Bagging_ con una diferencia: utiliza además una selección aleatoria de $m$ predictores para cada una de las muestras.

## **Boosting**

Como se mencionó anteriormente, la idea de este método es ajustar secuencialmente modelos débiles. De esta manera, cada nuevo modelo utiliza la información de los modelos anteriores para corregir sus propios errores. A diferencia del _bagging_, no se realizan muestreos aleatorios.  
Las implementaciones de este método se caracterizan por tener muchos hiperparámetros asociados, entre los cuales los más importantes son:

- **Número de _weak learners_**: Este parámetro determina cuántas iteraciones se realizarán. A diferencia del _bagging_, en el _boosting_ puede generarse _overfitting_ si se realizan demasiadas iteraciones.
- **Learning Rate**: Controla la influencia que tiene cada _weak learner_ en el conjunto total. En otras palabras, determina la velocidad a la que aprende el modelo. Generalmente, se utiliza un valor entre 0.001 y 0.01. Si el valor es más bajo, se necesitarán más iteraciones, pero el riesgo de _overfitting_ será menor. Sin embargo, esto implica que el modelo tardará más en aprender.
- **Profundidad máxima del árbol permitido**: Limita la complejidad de cada árbol en el modelo.

### **Extreme Gradient Boosting**

XGBoost (por sus siglas en _eXtreme Gradient Boosting_) es uno de los algoritmos de _ensembles_ más populares, especialmente en competencias de _Machine Learning_ en sitios como Kaggle. Este _ensemble_ proviene de la familia de algoritmos _Boosting_. XGBoost es un modelo de _boosting_ que, además de utilizar dicha técnica, optimiza la función de pérdida mediante el descenso por gradiente. Para entender este algoritmo, es necesario explicar el concepto de descenso por gradiente. Todos los algoritmos de _Gradient Boosting_ deben tener tres elementos:

- Una función de pérdida a optimizar.
- Un algoritmo base débil (por ejemplo, árboles de decisión).
- Un modelo aditivo para sumar cada uno de los estimadores o algoritmos base.

Los árboles se agregan de a uno, y los que ya fueron agregados no cambian. Para definir los parámetros de los árboles, se utiliza el concepto de descenso por gradiente para minimizar la función de pérdida. A continuación, veremos el método de optimización del descenso por gradiente.

#### Métodos de optimización

**Estimación de Máxima Verosimilitud**: Como se mencionó anteriormente, la idea del método es encontrar el mejor valor de estimación para un parámetro, de manera que se maximice la probabilidad de obtener una muestra determinada. Sabiendo que la probabilidad de un suceso es $P(X) = p \times (1 − p)$, veamos un ejemplo:

Lanzaremos una moneda 5 veces y observaremos su secuencia: Seca, Cara, Cara, Cara, Seca.

$$P(\text{Seca, Cara, Cara, Cara, Seca}) = (1 − p) \times p \times p \times p \times (1 − p) = p^3 \times (1 − p)^2$$

Con esta secuencia, el $p$ de Cara es 0.6 ($\frac{3}{5}$) y, por ende, la probabilidad es:

$$P(\text{Seca, Cara, Cara, Cara, Seca}) = 0.6^3 \times (1 − 0.6)^2 = 0.0346$$

Para esta muestra, calcularemos todos los valores posibles de $p$ y sus respectivas probabilidades.

Haciendo esta comparación, observamos que para esta muestra de observaciones, el valor donde se encuentra la mayor probabilidad es cuando $p=0.6$. Entonces, el parámetro $\hat{p}$ será 0.6. Este ejemplo nos muestra que el parámetro estimado por máxima verosimilitud es igual a la frecuencia relativa del evento que deseamos estimar. Ahora intentaremos generalizar la fórmula y verificar si esta experimentación es correcta.

$$\text{Prob}(k \text{ caras en } n \text{ lanzamientos}) = p^k \times (1-p)^{n-k} = L(p)$$

La función $L(p)$ es conocida como la función de verosimilitud y puede definirse como la función de densidad conjunta de la muestra observada, pero considerada como una función del parámetro.


## Parámetros Boosting
El boosting es un ensamble de *learners* débiles (generalmente árboles de decisión) que trabajan en conjunto para poder lograr resultados mucho más robustos en su conjunto.
La forma en que se generan es secuencial, es decir, se genera un árbol de, el mismo es evaluado, luego se generan *pesos* en cada una de las observaciones, las observaciones que fueron érroneas en sus predicciones, se les dará mayor peso, las que fueron acertadas, menor. Esto se hace para que el árbo siguiente haga más hincapié en aquellas observaciones que fueron mal predichas.

Existen varias implementaciones de este algoritmo:

* XGBoost (Extreme Gradient Boosting): Es uno de los algoritmos que estuvo dominando las competencias de data mining a nivel global. Creado por Tianqui Chen y Carlos Guestrin, logra muy buenos resultados y excelentes generalizaciones. En el siguiente [link](https://xgboost.readthedocs.io/en/latest/) podrán encontrar toda su documentación.
* LightGBM: Es una implementación de un modelo boosting creado por Microsoft, mejora el XGBoost generando muy buenos resultados. Para ver la documentación de dicha implementación realizar click en el siguiente [enlace](https://lightgbm.readthedocs.io/en/latest/pythonapi/lightgbm.LGBMClassifier.html#lightgbm.LGBMClassifier)

Los parámetros para ajustar del **XGBoost** son:

* base_score: Este parámetro es el score que se le asigna inicialmente a todas las predicciones.
* Booster: Este parámetro nos indica que se desea usar como *learner* base. Las tres opciones de este algoritmo son: gbtree, dart (ambos se basan en árboles) y gblinear (funciones lineales)
* colsample_bylevel: Se realiza un muestreo de las columnas a elegir encada uno de los niveles del árbol. Es decir, cada vez que crece un nivel el árbol, se elije una característica de una submuestra.
* colsample_bynode: Por cada nodo del árbol se realiza una muestra de las posibles variables para elegir y realizar un nuevo corte.
* colsample_bytree: Por cada árbol nuevo, se realiza una selección de variables.
* gamma: Es igual al min_split_loss. Es la mínima reducción de pérdidas que se requiere para poder realizar un nuevo split en un nodo hoja. Cuanto más alto sea el valor de gamma, más conservador será el árbol.
* importance_type: Es el tipo de feature importance. Los valores son gain, weight o total_gain.
* learning_rate: Es la tasa de aprendizaje. Si el valor es muy chico, los pasos para buscar los mínimos de las funciones serán más cortos, si el valor es muy grande, los pasos serán más largos.
* max_delta_step: Es el máximo delta que se puede tener cuando se estimen los pesos de los árboles.
* max_depth: Profundidad máxima de los árboles bases.
* missing: El valor con el que se completarán los valores faltantes.
* n_estimators: Cantidad de *learners* base que tendrá el modelo final.
* n_jobs: Número de jobs (threads) que correrán en paralelo.
* random_state: Valor de la semilla para que puedan ser reproducidos los resultados.
* reg_alpha: Regularización L1.
* reg_lambda: Regularización L2.
* scale_pos_weight: Balanceo entre los pesos de las clases positivas o negativas.
* subsample: Es un ratio de muestreo para las instancias de entrenamiento.
* verbosity: Valor por default=1. Este parámetro nos indicará que mensajes se mostrarán por pantalla. El valor 0 indica falta de mensajes, 1 solo advertencias, 2 información, 3 información de debug.

# Optimizaciones

## 1) Búsqueda Aleatoria de Paránetros con CV estratificado (AUC) - Usando GPU

```python
# ====== AFINADOR RÁPIDO DE XGBOOST PARA COLAB (con slicing seguro) ======
import gc, json
import numpy as np
import pandas as pd
import xgboost as xgb
from xgboost.callback import EarlyStopping
from sklearn.model_selection import StratifiedKFold, train_test_split
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# --- Utilidades ---
def binarize_y(y):
    """Convierte etiquetas 'NO'/'SI' a 0/1 de forma robusta."""
    y = pd.Series(y).astype(str).str.strip().str.upper()
    m = {'NO': 0, 'SI': 1}
    yb = y.map(m)
    if yb.isna().any():
        raise ValueError(f"Etiquetas desconocidas en y: {sorted(y[yb.isna()].unique())}")
    return yb.astype(np.int8)

def to_float32_array(X):
    """Deja solo columnas numéricas y castea a float32 para ahorrar RAM/VRAM."""
    if hasattr(X, "select_dtypes"):
        Xn = X.select_dtypes(include=[np.number]).copy()
        for c in Xn.columns:
            if Xn[c].dtype != np.float32:
                Xn[c] = Xn[c].astype(np.float32)
        return Xn.values
    return np.asarray(X, dtype=np.float32)

# --- Datos binarizados y en float32 ---
y_train_bin = binarize_y(y_train)
y_test_bin  = binarize_y(y_test)
X_train_np  = to_float32_array(X_train)
X_test_np   = to_float32_array(X_test)

# --- Detección de GPU (usamos GPU solo si existe) ---
USE_GPU = False
try:
    import cupy as cp
    USE_GPU = cp.cuda.runtime.getDeviceCount() > 0
except Exception:
    USE_GPU = False

# --- Matriz de entrenamiento preconstruida y "sliceable" para CV ---
DMX = xgb.DMatrix
dall = DMX(X_train_np, label=y_train_bin)

# --- Espacio de búsqueda pequeño (rápido/seguro en Colab) ---
pos = int(y_train_bin.sum()); neg = int(len(y_train_bin) - pos)
spw = max(neg / max(pos, 1), 1.0)  # manejo de desbalance

search_space = {
    "max_depth":        [3, 4, 5, 6],
    "min_child_weight": [1, 2, 4, 6],
    "subsample":        [0.7, 0.85, 1.0],
    "colsample_bytree": [0.7, 0.85, 1.0],
    "gamma":            [0.0, 0.1, 0.5],  # = min_split_loss
    "reg_alpha":        [0.0, 1e-2, 0.1],
    "reg_lambda":       [0.5, 1.0, 2.0],
    "scale_pos_weight": [1.0, round(spw, 2)],
}
rng = np.random.RandomState(42)

def sample_params(space, rnd):
    """Muestrea y convierte a tipos nativos de Python para evitar numpy scalars."""
    out = {}
    for k, vals in space.items():
        v = rnd.choice(vals)
        # numpy scalar? pásalo a nativo
        if hasattr(v, "item"):
            v = v.item()
        out[k] = v
    return out

# --- Configuración de CV y entrenamiento (rápido) ---
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
NUM_BOOST_ROUND = 1200
ES_ROUNDS       = 60
ETA             = 0.08

# En XGBoost 3.x: device='cuda' + tree_method='hist' => entrenamiento en GPU.
# Si no hay GPU, usamos device='cpu'; 'sampling_method' solo vale en GPU.
base_params = {
    "objective": "binary:logistic",
    "eval_metric": "auc",
    "tree_method": "hist",
    "device": "cuda" if USE_GPU else "cpu",
    "eta": ETA,
    "max_bin": 256,  # bajar a 128 si hay poca VRAM
    "nthread": -1
}
if USE_GPU:
    base_params["sampling_method"] = "gradient_based"

def cv_auc(params):
    """Entrena por fold con early stopping y devuelve el AUC medio."""
    scores = []
    for tr_idx, va_idx in cv.split(np.zeros(len(y_train_bin)), y_train_bin):
        dtr = dall.slice(tr_idx)   # DMatrix soporta slicing
        dva = dall.slice(va_idx)

        booster = xgb.train(
            params={**base_params, **params},
            dtrain=dtr,
            num_boost_round=NUM_BOOST_ROUND,
            evals=[(dva, "valid")],
            callbacks=[EarlyStopping(rounds=ES_ROUNDS, save_best=True, maximize=True)],
            verbose_eval=False
        )
        preds = booster.predict(dva)
        scores.append(roc_auc_score(dva.get_label(), preds))

        del booster, dtr, dva
        gc.collect()
    return float(np.mean(scores))

# --- Búsqueda aleatoria (pocas pruebas para no romper Colab) ---
N_TRIALS = 12
mejores_params, mejor_auc_cv = None, -np.inf
for _ in range(N_TRIALS):
    p = sample_params(search_space, rng)
    s = cv_auc(p)
    if s > mejor_auc_cv:
        mejor_auc_cv, mejores_params = s, p

print("Mejor AUC de CV:", round(mejor_auc_cv, 4))
print("Mejores hiperparámetros:", mejores_params)

# --- Reentrenamiento final con early stopping sobre un holdout pequeño ---
idx = np.arange(len(y_train_bin))
tr_idx, va_idx = train_test_split(idx, test_size=0.15, stratify=y_train_bin, random_state=42)
dtr_full = dall.slice(tr_idx)
dval     = dall.slice(va_idx)

final_params = {**base_params, **mejores_params}
final_booster = xgb.train(
    params=final_params,
    dtrain=dtr_full,
    num_boost_round=2000,
    evals=[(dval, "valid")],
    callbacks=[EarlyStopping(rounds=100, save_best=True, maximize=True)],
    verbose_eval=False
)

# --- Evaluación en test + Curva ROC ---
dtest = DMX(X_test_np, label=y_test_bin)
y_score = final_booster.predict(dtest)
auc_test = roc_auc_score(y_test_bin, y_score)
fpr, tpr, _ = roc_curve(y_test_bin, y_score, drop_intermediate=False)
print(f"AUC en test = {auc_test:.4f}")

fig, ax = plt.subplots(figsize=(7,7))
ax.plot(fpr, tpr, lw=2, label=f'ROC (AUC = {auc_test:.3f})')
ax.plot([0,1], [0,1], '--', lw=1)
ax.set(xlim=(0,1), ylim=(0,1), xlabel='FPR', ylabel='TPR', title='Gráfico ROC para el modelo')
ax.legend(loc='lower right'); ax.grid(True, alpha=0.3)
plt.show()

# --- Guardar modelo a archivo (y parámetros) ---
# 1) Guardar el booster en formato JSON (portátil entre versiones)
final_booster.save_model("modelo_xgb.json")
print("Modelo guardado en: modelo_xgb.json")

# 2) (Opcional) Guardar en formato UBJ (binario universal de XGBoost)
final_booster.save_model("modelo_xgb.ubj")
print("Modelo guardado en: modelo_xgb.ubj")

# 3) Guardar los mejores hiperparámetros usados (saneados a tipos nativos)
safe_params = {k: (v.item() if hasattr(v, "item") else v) for k, v in mejores_params.items()}
with open("mejores_hiperparametros.json", "w") as f:
    json.dump(safe_params, f, indent=2)
print("Hiperparámetros guardados en: mejores_hiperparametros.json")

# 4) (Opcional) Cómo cargar el modelo después
loaded = xgb.Booster()
loaded.load_model("modelo_xgb.json")
y_score_loaded = loaded.predict(dtest)  # mismo uso que final_booster

# Limpieza de objetos pesados
del dall, dtr_full, dval, dtest
gc.collect()
```

**Por qué esto ayuda**

* AUC como eval_metric + early stopping: entrena directamente para ordenar bien las probabilidades (lo que maximiza ROC-AUC) y evita sobreajuste.

* scale_pos_weight: corrige el sesgo cuando hay desequilibrio (usa neg/pos como punto de partida).

* Búsqueda aleatoria con CV: explora combinaciones clave que más afectan el ranking de scores: max_depth, min_child_weight, gamma, subsample, colsample_bytree, reg_alpha, reg_lambda.

```python
# ==== Cargar modelo XGBoost y graficar ROC con GPU si existe ====
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
import xgboost as xgb
from sklearn.metrics import roc_auc_score, roc_curve

# --- Helpers (por si no existen X_test_np / y_test_bin) ---
def binarize_y(y):
    y = pd.Series(y).astype(str).str.strip().str.upper()
    m = {'NO': 0, 'SI': 1}
    yb = y.map(m)
    if yb.isna().any():
        raise ValueError(f"Etiquetas desconocidas en y_test: {sorted(y[yb.isna()].unique())}")
    return yb.astype(np.int8)

def to_float32_array(X):
    if hasattr(X, "select_dtypes"):
        Xn = X.select_dtypes(include=[np.number]).copy()
        for c in Xn.columns:
            if Xn[c].dtype != np.float32:
                Xn[c] = Xn[c].astype(np.float32)
        return Xn.values
    return np.asarray(X, dtype=np.float32)

# --- Preparar X_test e y_test (reusar si ya existen) ---
try:
    X_te = X_test_np
except NameError:
    if 'X_test' not in globals():
        raise RuntimeError("No encuentro X_test. Definí X_test o X_test_np antes de ejecutar.")
    X_te = to_float32_array(X_test)

try:
    y_te = y_test_bin
except NameError:
    if 'y_test' not in globals():
        raise RuntimeError("No encuentro y_test. Definí y_test o y_test_bin antes de ejecutar.")
    y_te = binarize_y(y_test)

# --- 1) Cargar el modelo guardado ---
if not os.path.exists("modelo_xgb.json"):
    raise FileNotFoundError("No se encontró 'modelo_xgb.json'. Ejecutá el entrenamiento/guardado antes.")
loaded = xgb.Booster()
loaded.load_model("modelo_xgb.json")

# --- 2) Predicción (GPU si disponible, si no CPU) ---
used_gpu = False
y_score = None
# Intento de ruta GPU con CuPy + inplace_predict
try:
    import cupy as cp
    if cp.cuda.runtime.getDeviceCount() > 0:
        X_te_gpu = cp.asarray(X_te)                  # mover features a GPU
        y_score_gpu = loaded.inplace_predict(X_te_gpu)  # inferencia en GPU
        y_score = cp.asnumpy(y_score_gpu)            # volver a NumPy para métricas sklearn
        used_gpu = True
    else:
        raise RuntimeError("GPU no detectada por CuPy.")
except Exception:
    # Fallback CPU: inplace_predict (rápido) o predict con DMatrix
    try:
        y_score = loaded.inplace_predict(X_te)       # CPU inplace
    except Exception:
        dtest = xgb.DMatrix(X_te, label=y_te)        # alternativa universal
        y_score = loaded.predict(dtest)
    used_gpu = False

print("Inferencia usando GPU:", "Sí" if used_gpu else "No")

# --- 3) Curva ROC + AUC ---
auc_test = roc_auc_score(y_te, y_score)
fpr, tpr, _ = roc_curve(y_te, y_score, drop_intermediate=False)
print(f"AUC en test (modelo cargado) = {auc_test:.4f}")

fig, ax = plt.subplots(figsize=(7,7))
ax.plot(fpr, tpr, lw=2, label=f'ROC (AUC = {auc_test:.3f})')
ax.plot([0,1], [0,1], '--', lw=1)  # línea base
ax.set(xlim=(0,1), ylim=(0,1), xlabel='FPR', ylabel='TPR', title='Curva ROC - Modelo cargado')
ax.legend(loc='lower right'); ax.grid(True, alpha=0.3)
plt.show()
```

## 2) Inferencia XGBoost con GPU automática, carga flexible y probabilidades

Implementación robusta:

*   Intenta primero usar el modelo ya creado en memoria (loaded, final_booster o xgbModel.get_booster()).

*   Si no lo encuentra, carga modelo_xgb.json.

*   Detecta GPU y usa CuPy + inplace_predict si está disponible; si no, cae a CPU.

*   Genera xbgProbs y ejecuta los mini chequeos útiles.

*   Calcula AUC si y_test_bin está disponible.

*   Grafica ROC si se desea.

```python
# === Predicción con modelo ya en memoria o, si no, desde archivo; GPU si está disponible ===
import os, numpy as np, pandas as pd
import xgboost as xgb

# -------- Helpers mínimos --------
def binarize_y(y):
    y = pd.Series(y).astype(str).str.strip().str.upper()
    m = {'NO': 0, 'SI': 1}
    yb = y.map(m)
    if yb.isna().any():
        raise ValueError(f"Etiquetas desconocidas en y_test: {sorted(y[yb.isna()].unique())}")
    return yb.astype(np.int8)

def to_float32_array(X):
    if hasattr(X, "select_dtypes"):
        Xn = X.select_dtypes(include=[np.number]).copy()
        for c in Xn.columns:
            if Xn[c].dtype != np.float32:
                Xn[c] = Xn[c].astype(np.float32)
        return Xn.values
    return np.asarray(X, dtype=np.float32)

# -------- Preparación de X_test / y_test (reusar si ya existen) --------
try:
    X_te = X_test_np
except NameError:
    if 'X_test' not in globals():
        raise RuntimeError("No se encontró X_test ni X_test_np en el entorno.")
    X_te = to_float32_array(X_test)

try:
    y_te = y_test_bin
except NameError:
    if 'y_test' in globals():
        y_te = binarize_y(y_test)
    else:
        y_te = None  # si no está, se continúa sin calcular AUC

# -------- Resolver modelo: primero en memoria, si no, desde archivo --------
booster = None
modelo_origen = None

if 'loaded' in globals():
    booster = loaded
    modelo_origen = "loaded (Booster en memoria)"
elif 'final_booster' in globals():
    booster = final_booster
    modelo_origen = "final_booster (Booster en memoria)"
elif 'xgbModel' in globals():
    try:
        booster = xgbModel.get_booster()
        modelo_origen = "xgbModel.get_booster() (desde wrapper sklearn)"
    except Exception:
        booster = None

if booster is None:
    if os.path.exists("modelo_xgb.json"):
        booster = xgb.Booster()
        booster.load_model("modelo_xgb.json")
        modelo_origen = "archivo modelo_xgb.json"
    else:
        raise FileNotFoundError("No se halló un modelo en memoria ni el archivo 'modelo_xgb.json'.")

print("Modelo utilizado:", modelo_origen)

# -------- Predicción: primero GPU (CuPy + inplace_predict), si no CPU --------
use_gpu = False
xbgProbs = None
origen_inferencia = None

try:
    import cupy as cp
    use_gpu = cp.cuda.runtime.getDeviceCount() > 0
except Exception:
    cp = None
    use_gpu = False

if use_gpu:
    try:
        X_te_gpu = cp.asarray(X_te)
        y_pred_gpu = booster.inplace_predict(X_te_gpu)  # GPU
        xbgProbs = cp.asnumpy(y_pred_gpu)               # volver a NumPy
        origen_inferencia = "GPU (CuPy + inplace_predict)"
    except Exception:
        # Fallback CPU si algo falla en la ruta GPU
        try:
            xbgProbs = booster.inplace_predict(X_te)    # CPU inplace
            origen_inferencia = "CPU (inplace_predict)"
        except Exception:
            dtest = xgb.DMatrix(X_te)
            xbgProbs = booster.predict(dtest)
            origen_inferencia = "CPU (DMatrix + predict)"
else:
    # Ruta CPU directa
    try:
        xbgProbs = booster.inplace_predict(X_te)
        origen_inferencia = "CPU (inplace_predict)"
    except Exception:
        dtest = xgb.DMatrix(X_te)
        xbgProbs = booster.predict(dtest)
        origen_inferencia = "CPU (DMatrix + predict)"

print("Inferencia usando:", origen_inferencia)

# -------- Mini chequeos útiles --------
print("----- Chequeos de salida -----")
print("Tipo de xbgProbs:", type(xbgProbs).__name__)
print("Shape:", np.shape(xbgProbs))
print("Primeros 5:", np.asarray(xbgProbs)[:5])
print("Mín / Máx:", float(np.min(xbgProbs)), float(np.max(xbgProbs)))

fuera_de_rango = np.logical_or(xbgProbs < 0.0, xbgProbs > 1.0)
if np.any(fuera_de_rango):
    idx = np.where(fuera_de_rango)[0][:10]
    print("Advertencia: se hallaron probabilidades fuera de [0,1] en índices:", idx)
else:
    print("Las probabilidades se encuentran dentro de [0, 1].")

# -------- (Opcional) AUC rápido si y_test_bin / y_te está disponible --------
try:
    if y_te is not None:
        from sklearn.metrics import roc_auc_score
        auc = roc_auc_score(y_te, xbgProbs)
        print(f"AUC (modelo usado) = {auc:.4f}")
except Exception:
    pass

# -------- (Opcional) Curva ROC --------

try:
     from sklearn.metrics import roc_curve
     import matplotlib.pyplot as plt
     fpr, tpr, _ = roc_curve(y_te, xbgProbs, drop_intermediate=False)
     fig, ax = plt.subplots(figsize=(7,7))
     ax.plot(fpr, tpr, lw=2, label=f'ROC (AUC = {auc:.3f})' if y_te is not None else 'ROC')
     ax.plot([0,1], [0,1], '--', lw=1)
     ax.set(xlim=(0,1), ylim=(0,1), xlabel='FPR', ylabel='TPR', title='Curva ROC - Modelo en uso')
     ax.legend(loc='lower right'); ax.grid(True, alpha=0.3)
     plt.show()
except Exception:
     pass
```

## Calculo el _accuracy_ del modelo

Idea general

Entrenar un LGBMClassifier.

* Obtener probabilidades de la clase positiva.

* Evitar errores de tipo al calcular métricas (coherencia entre 'NO'/'SI' y 0/1).

* un umbral óptimo con Youden (J)

* Reportar accuracy, matriz de confusión, y curva ROC/AUC.

image.png

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve

# 1) Asegurar que tomamos la probabilidad de la clase positiva 'SI'
idx_pos = list(lgbModel.classes_).index('SI')  # evita asumir que es la última columna
lgbProbs = lgbModel.predict_proba(X_test)[:, idx_pos]  # P(y='SI')

# ---------- Opción A: Mantener etiquetas 'NO'/'SI' ----------
y_pred_labels = np.where(lgbProbs >= 0.5, 'SI', 'NO')
acc = accuracy_score(y_test, y_pred_labels)
print(f"Accuracy (umbral 0.5, etiquetas string): {acc:.4f}")

# (Opcional) otros resúmenes usando etiquetas string
print(classification_report(y_test, y_pred_labels))
print(confusion_matrix(y_test, y_pred_labels))

# ---------- Opción B: Binarizar y comparar con 0/1 ----------
y_test_bin = (
    pd.Series(y_test).astype(str).str.strip().str.upper().map({'NO': 0, 'SI': 1}).astype(int)
)
y_pred_bin = (lgbProbs >= 0.5).astype(int)
acc_bin = accuracy_score(y_test_bin, y_pred_bin)
print(f"Accuracy (umbral 0.5, binario): {acc_bin:.4f}")

# ---------- (Opcional) Elegir umbral óptimo por Youden (TPR - FPR) ----------
fpr, tpr, thresholds = roc_curve(y_test_bin, lgbProbs)
j_scores = tpr - fpr
thr_opt = thresholds[j_scores.argmax()]
y_pred_opt = (lgbProbs >= thr_opt).astype(int)
acc_opt = accuracy_score(y_test_bin, y_pred_opt)
print(f"Umbral óptimo ≈ {thr_opt:.4f} | Accuracy con umbral óptimo: {acc_opt:.4f}")
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score, precision_recall_curve, classification_report, confusion_matrix
import numpy as np

# 1) Métricas de ranking
roc_auc = roc_auc_score(y_test_bin, lgbProbs)
pr_auc  = average_precision_score(y_test_bin, lgbProbs)
prev    = y_test_bin.mean()
print(f"ROC-AUC: {roc_auc:.4f} | PR-AUC: {pr_auc:.4f} | Prevalencia SI: {prev:.4%}")

def eval_at(thr, name=""):
    pred = (lgbProbs >= thr).astype(int)
    print(f"\n--- {name} thr={thr:.4f} ---")
    print(classification_report(y_test_bin, pred, target_names=["NO","SI"], digits=4))
    print(confusion_matrix(y_test_bin, pred))

# Re-evalúa en 0.5 y en thr_opt (que ya calculaste)
eval_at(0.5, "Umbral 0.5")
eval_at(thr_opt, "Umbral Youden")

# 2) Elegir umbral por costo (ajusta C_FP y C_FN a tu caso)
C_FP, C_FN = 1, 10   # ejemplo: FN cuesta 10x un FP
thr_cost = C_FP / (C_FP + C_FN)   # umbral Bayesiano con probs calibradas
eval_at(thr_cost, f"Umbral costo (C_FP={C_FP}, C_FN={C_FN})")

# 3) Alternativa operativa: Top-K (p.ej., revisar 10k casos)
K = 10_000
order = np.argsort(lgbProbs)[::-1]
pred_topk = np.zeros_like(y_test_bin.values)
pred_topk[order[:K]] = 1
print(f"\n--- Top-{K} ---")
print(classification_report(y_test_bin, pred_topk, target_names=["NO","SI"], digits=4))
print(confusion_matrix(y_test_bin, pred_topk))
```
